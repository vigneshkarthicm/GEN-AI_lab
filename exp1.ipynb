```python
from transformers import pipeline

print("Loading model...")
generator = pipeline("text-generation", model="distilgpt2")
print("Ready!")
```

```python
prompt=input("Enter your prompt: ")

output=generator(
    prompt,
    max_new_tokens=150,
    temperature=0.7,
    do_sample=True
)

print(output[0]["generated_text"])
```

```python
while True:
    text=input("\nPrompt (exit to quit): ")
    if text.lower()=="exit":
        break

    output=generator(
        text,
        max_new_tokens=150,
        temperature=0.7,
        do_sample=True
    )

    print("\nGenerated Text:\n")
    print(output[0]["generated_text"])
    print("-"*80)
```

```python
def generate_text(prompt,
                  max_tokens,
                  temperature):

    output = generator(
        prompt,
        max_new_tokens=max_tokens,
        temperature=temperature,
        do_sample=True,
        pad_token_id=generator.tokenizer.eos_token_id
    )

    return output[0]["generated_text"]
```

```python
import gradio as gr

with gr.Blocks(theme=gr.themes.Soft()) as demo:

    gr.Markdown(
        """
        # 🤖 AI Content Generator

        ### Transformer Model : DistilGPT-2
        """
    )

    with gr.Row():

        with gr.Column():

            prompt = gr.Textbox(
                label="Enter Prompt",
                lines=6,
                placeholder="Example: Explain Artificial Intelligence..."
            )

            max_tokens = gr.Slider(
                minimum=50,
                maximum=500,
                value=200,
                step=10,
                label="Max New Tokens"
            )

            temperature = gr.Slider(
                minimum=0.1,
                maximum=1.5,
                value=0.7,
                step=0.1,
                label="Temperature"
            )

            generate = gr.Button(
                "Generate",
                variant="primary"
            )

        with gr.Column():

            output = gr.Textbox(
                label="Generated Content",
                lines=20
            )

    generate.click(
        fn=generate_text,
        inputs=[
            prompt,
            max_tokens,
            temperature
        ],
        outputs=output
    )

demo.launch(share=True)
```
